```python
import google.generativeai as genai
from google.colab import userdata

GOOGLE_API_KEY=userdata.get('aaditi')
genai.configure(api_key=GOOGLE_API_KEY)
```

```python
gemini_model=genai.GenerativeModel("gemini-3.5-flash")
prompt_text="I want to open an Indian game show. Suggest Five creative names for it"
response = gemini_model.generate_content(prompt_text)
print(response.text)
```

```python
!pip install -U langchain-groq
!pip install -U langchain

from google.colab import userdata
from langchain_groq import ChatGroq

GROQ_API_KEY = userdata.get('aaditi21')

groq_model = ChatGroq(model="openai/gpt-oss-20b",groq_api_key=GROQ_API_KEY)
prompt_text = "I want to open an Indian game show. Suggest five creative names."
response = groq_model.invoke(prompt_text)
print(response.content)
```

```python
!pip install -U langchain-google-genai

from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

GOOGLE_API_KEY = userdata.get("aaditi")
#llm implementation
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    google_api_key=GOOGLE_API_KEY
)

from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

restaurant_name_prompt = PromptTemplate(
    input_variables=["cuisine"],
    template="I want to open a restaurant for {cuisine} food. Suggest a fancy name for this."
)

name_chain = restaurant_name_prompt | llm | StrOutputParser()
print("Suggested Indian Restaurant Names:")
response_name = name_chain.invoke({"cuisine": "Indian"})
print(response_name)
```
